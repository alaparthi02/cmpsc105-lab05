## Plot 1

**Scenario**: A researcher wants to show that Gentoo penguins are significantly heavier than other species, but their plot is visually overwhelming and exaggerates the differences.

```python
penguins = sns.load_dataset("penguins")

ax = sns.barplot(data=penguins, x="species", y="body_mass_g", palette="hsv")
ax.set_ylim(0, 5500) 
plt.title("Penguin Mass by Species")
plt.show()
```

Issues with the plot: The range of the y-axis was too small, which made the two other kinds of penguins appear much smaller compared to the Gentoo penguins. 


How you changed the plot and why: We increased the range of the y-axis and made the y-axis start at 0. This makes the differences between the species more accurately represented. The Gentoo penguins still obviously have a higher avg body mass than the other two kinds of penguins, but the graph no longer exaggerates this difference.


## Plot 2

**Scenario**: A student plotted the carat size versus price for 54,000 diamonds to see where the bulk of the market sits, but the result is just a solid block of ink.

```python
diamonds = sns.load_dataset("diamonds")
diamonds_sample = diamonds.sample(500)
sns.scatterplot(data=diamonds_sample, x="carat", y="price", color="blue", s=10)
plt.title("Diamond Price vs. Carat")
plt.show()
```

Issues with the plot: The size of each plot was too large and the graph displayed all 54000 diamonds, which caused the points to overlap and made the graph look very crowded, and hard to read.


How you changed the plot and why: I changed the plot to display a random sample of 500 diamonds and decreased the size of each plot. This makes the graph easier to read while still showing that the majority of the market is between approx 0.2 and 1 carat.


## Plot 3

The aviation board wants to see passenger growth month-over-month for each year in the 1950s, but the lines are indistinguishable.

```python
flights = sns.load_dataset("flights")
sns.lineplot(data=flights, x="month", y="passengers", units="year", estimator=None, hue="year")
plt.title("Passengers over Time")
plt.show()
```

Issues with the plot: There are too many lines with the same color, making them difficult to differentiate. The overlapping lines make it hard to clearly compare the passenger trends for specific years.

How you changed the plot and why: Without the legend, we would not be able to tell the years apart. To make it even more easier to read, we made the hues of the lines different so it is easier to tell the lines apart from one another.


## Plot 4

**Scenario**: A restaurant manager receives this chart showing the relationship between total bills and tips, split by smoking preference. 





```python
tips = sns.load_dataset("tips")

tips_sample = tips.sample(50)
sns.lmplot(
    data=tips_sample,
    x="total_bill",
    y="tip",
    hue="smoker",
    scatter_kws={"s": 10}
)
plt.show()
```

Issues with the plot: There is no specific title, so at first glance, it is difficult to understand what this is about exactly. Moreover, the data looks heavily concentrated (points are overlapping one another) which makes interpretation more difficult. 

How you changed the plot and why: To make the data more understandable, we took a smaller sample size of tips and made the plot size smaller. This made the data more scattered by spacing them out.
